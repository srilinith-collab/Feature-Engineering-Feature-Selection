## Question 1: What is Feature Engineering?

### Answer

Feature Engineering is the process of creating new features or transforming existing features from raw data so that Machine Learning models can learn useful patterns more effectively.

It uses existing information to create meaningful representations for the ML model.

For example, in the Titanic dataset:

`FamilySize = SibSp + Parch + 1`

This combines the number of siblings/spouses and parents/children with the passenger to represent the total family size.

Feature Engineering is important because the quality and representation of features can strongly affect Machine Learning performance.

```python
import pandas as pd
import numpy as np

# Load the Titanic dataset
df = pd.read_csv("Titanic-Dataset.csv")

print("Dataset loaded successfully.")
print("Dataset Shape:", df.shape)
print("\nFirst 5 Rows:")
print(df.head())
```

## Question 2: What is the difference between Feature Engineering and Feature Selection?

### Answer

**Feature Engineering** creates new features or transforms existing features to provide useful information to a Machine Learning model.

**Feature Selection** selects the most useful features from the available features and removes irrelevant or redundant features.

### Example

Feature Engineering:
`FamilySize = SibSp + Parch + 1`

Feature Selection:
Selecting `Age`, `Fare`, `Pclass`, and `FamilySize` while removing `PassengerId` because it is only an identifier.

In simple terms:

**Feature Engineering = Create or transform features**

**Feature Selection = Choose useful features**

## Question 3: Why is Domain Knowledge important in Feature Engineering?

### Answer

Domain Knowledge helps us understand what the raw data represents and which new features may be meaningful for the problem.

Without domain knowledge, we may create many features that are mathematically possible but have little practical value.

For example, in an e-commerce system:

`Average_Order_Value = Total_Spending / Number_of_Orders`

This feature has a clear business meaning because it represents the average amount spent per order.

Domain knowledge helps us create features that are meaningful, useful, interpretable, and relevant to the business problem.

```python
ecommerce = pd.DataFrame({
    "CustomerID": [101, 102, 103],
    "Total_Spending": [5000, 3000, 8000],
    "Number_of_Orders": [10, 5, 16]
})

ecommerce["Average_Order_Value"] = (
    ecommerce["Total_Spending"] / ecommerce["Number_of_Orders"]
)

print(ecommerce)
```

## Question 4: What is Feature Leakage?

### Answer

Feature Leakage occurs when information that should not be available at prediction time is used as an input feature.

This can make the model appear highly accurate during training or testing, but the model will perform poorly on real unseen data.

A common example is using the target variable directly to create a feature.

In the Titanic dataset, using `Survived` to create another feature such as:

`Survival_Info = Survived`

would be target leakage because `Survived` is the value that the model is supposed to predict.

### Key Rule

A feature should use only information that would be available at the time of prediction.

```python
import pandas as pd

df = pd.read_csv("Titanic-Dataset.csv")

# Correct feature: uses only input columns
df["FamilySize"] = df["SibSp"] + df["Parch"] + 1
df["IsAlone"] = (df["FamilySize"] == 1).astype(int)

print(df[["SibSp", "Parch", "FamilySize", "IsAlone"]].head())
```

## Question 5: What is an Interaction Feature?

### Answer

An Interaction Feature combines two or more existing features to represent their relationship.

For example, in the Titanic dataset:

`Age_Pclass_Interaction = Age × Pclass`

This represents the combined effect of passenger age and passenger class.

Interaction features can help a Machine Learning model learn relationships that may not be captured when features are considered separately.

However, creating too many interaction features can increase complexity and may cause overfitting.

```python
import pandas as pd

df = pd.read_csv("Titanic-Dataset.csv")

df["Age"] = df["Age"].fillna(df["Age"].median())
df["Age_Pclass_Interaction"] = df["Age"] * df["Pclass"]

print(df[["Age", "Pclass", "Age_Pclass_Interaction"]].head())
```

## Question 6: What is Binning?

### Answer

Binning is the process of converting continuous numerical values into meaningful groups or categories.

For example, passenger age can be grouped into:

- Child
- Teenager
- Young Adult
- Adult
- Senior

Binning can make numerical information easier to understand and can help represent broad non-linear patterns.

However, binning may result in some loss of detailed numerical information.

```python
import pandas as pd
import numpy as np

df = pd.read_csv("Titanic-Dataset.csv")

bins = [0, 12, 17, 30, 50, np.inf]
labels = ["Child", "Teenager", "Young Adult", "Adult", "Senior"]

df["Age_Group"] = pd.cut(df["Age"], bins=bins, labels=labels)

print(df[["Age", "Age_Group"]].head(10))
```

## Question 7: What is Feature Cardinality?

### Answer

Feature Cardinality refers to the number of unique values or categories in a feature.

For example:

- `Sex` has low cardinality.
- `Embarked` has low cardinality.
- `Ticket` has higher cardinality.
- `Name` has very high cardinality.

High-cardinality categorical features can create many columns when One-Hot Encoding is used.

For such features, techniques such as frequency encoding, target encoding, or rare-category grouping may be considered.

```python
import pandas as pd

df = pd.read_csv("Titanic-Dataset.csv")

columns = ["Sex", "Embarked", "Pclass", "Ticket", "Cabin"]

cardinality = pd.DataFrame({
    "Feature": columns,
    "Unique_Values": [df[col].nunique(dropna=True) for col in columns]
})

print(cardinality)
```

## Question 8: Explain Target Encoding.

### Answer

Target Encoding replaces each category with a value calculated from the target variable.

For a classification problem, the average target value for each category can be used.

For example, in the Titanic dataset, we can calculate the average `Survived` value for each `Sex`.

However, Target Encoding can cause data leakage if the target information is calculated using the complete dataset.

Therefore, the encoding values should be calculated using the training data only and then applied to the test data.

```python
import pandas as pd
from sklearn.model_selection import train_test_split

df = pd.read_csv("Titanic-Dataset.csv")

train, test = train_test_split(
    df, test_size=0.2, random_state=42, stratify=df["Survived"]
)

target_mean = train.groupby("Sex")["Survived"].mean()

train = train.copy()
test = test.copy()

train["Sex_Target_Encoded"] = train["Sex"].map(target_mean)
test["Sex_Target_Encoded"] = test["Sex"].map(target_mean)

print(train[["Sex", "Survived", "Sex_Target_Encoded"]].head())
```

## Question 9: Explain Frequency Encoding.

### Answer

Frequency Encoding replaces each category with the number of times that category occurs.

In the Titanic dataset, `Ticket` can be used for Frequency Encoding.

For example, if a ticket appears for three passengers, its frequency value will be 3.

This method is useful for high-cardinality categorical features because it does not create a separate column for every category.

In a real ML workflow, frequency values should be learned from the training data only.

```python
import pandas as pd

df = pd.read_csv("Titanic-Dataset.csv")

ticket_frequency = df["Ticket"].value_counts()
df["Ticket_Frequency"] = df["Ticket"].map(ticket_frequency)

print(df[["Ticket", "Ticket_Frequency"]].head(10))
```

## Question 10: When should Log Transformation be used?

### Answer

Log Transformation is mainly useful when a numerical feature is highly right-skewed.

It compresses large values and reduces the influence of extreme values.

A common transformation is:

`log1p(x) = log(1 + x)`

In the Titanic dataset, the `Fare` feature can be transformed using `log1p()`.

The transformation should be used after checking the distribution of the feature.

```python
import pandas as pd
import numpy as np

df = pd.read_csv("Titanic-Dataset.csv")

df["Fare_Log"] = np.log1p(df["Fare"])

print(df[["Fare", "Fare_Log"]].head())
```

## Question 11: What is the Curse of Dimensionality?

### Answer

The Curse of Dimensionality refers to problems that can occur when the number of features becomes very large.

As the number of dimensions increases:

- Data becomes more sparse.
- Computational cost increases.
- Distance-based methods can become less effective.
- More data may be required.
- The risk of overfitting can increase.

Feature Selection and Dimensionality Reduction can help reduce these problems.

```python
import pandas as pd

df = pd.read_csv("Titanic-Dataset.csv")

features = ["Pclass", "Age", "SibSp", "Parch", "Fare"]

print("Number of Features:", len(features))
print("Features:", features)
```

## Question 12: Explain PCA.

### Answer

PCA stands for Principal Component Analysis.

PCA is a dimensionality reduction technique that converts the original numerical features into a smaller number of new features called Principal Components.

The first principal component captures the maximum possible variance in the data.

PCA can be useful when there are many numerical features or when dimensionality needs to be reduced.

Before applying PCA, numerical features should generally be scaled.

One limitation is that Principal Components are combinations of original features, so they can be harder to interpret.

```python
import pandas as pd
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

df = pd.read_csv("Titanic-Dataset.csv")

features = ["Pclass", "Age", "SibSp", "Parch", "Fare"]

X = df[features]
X = SimpleImputer(strategy="median").fit_transform(X)
X = StandardScaler().fit_transform(X)

pca = PCA(n_components=2)
X_pca = pca.fit_transform(X)

print("Original Features:", len(features))
print("PCA Components:", X_pca.shape[1])
print("Explained Variance:", pca.explained_variance_ratio_)
```

## Question 13: What is Feature Importance?

### Answer

Feature Importance measures how useful a feature is for making predictions in a Machine Learning model.

Tree-based models such as Decision Trees and Random Forests can calculate feature importance.

A higher importance value means that the feature contributed more to the model's predictions.

However:

**Feature Importance shows predictive usefulness, not causation.**

A highly important feature does not necessarily cause the target outcome.

```python
import pandas as pd
from sklearn.ensemble import RandomForestClassifier

df = pd.read_csv("Titanic-Dataset.csv")

features = ["Pclass", "Age", "SibSp", "Parch", "Fare"]

X = df[features].copy()
X[["Age", "Fare"]] = X[["Age", "Fare"]].fillna(
    X[["Age", "Fare"]].median()
)

y = df["Survived"]

model = RandomForestClassifier(n_estimators=100, random_state=42)
model.fit(X, y)

importance = pd.DataFrame({
    "Feature": features,
    "Importance": model.feature_importances_
}).sort_values("Importance", ascending=False)

print(importance)
```

## Question 14: Create Five Useful Features for the Titanic Dataset.

### Answer

Five useful features for the Titanic dataset are:

1. **FamilySize** – total number of immediate family members travelling with the passenger.
2. **IsAlone** – identifies whether the passenger travelled alone.
3. **FarePerPerson** – represents the approximate fare per family member.
4. **AgeGroup** – converts age into meaningful age categories.
5. **Age_Pclass_Interaction** – represents the interaction between age and passenger class.

These features provide additional information that may help a Machine Learning model identify survival patterns.

```python
import pandas as pd
import numpy as np

df = pd.read_csv("Titanic-Dataset.csv")

df["FamilySize"] = df["SibSp"] + df["Parch"] + 1
df["IsAlone"] = (df["FamilySize"] == 1).astype(int)
df["FarePerPerson"] = df["Fare"] / df["FamilySize"]

bins = [0, 12, 17, 30, 50, np.inf]
labels = ["Child", "Teenager", "Young Adult", "Adult", "Senior"]
df["AgeGroup"] = pd.cut(df["Age"], bins=bins, labels=labels)

df["Age_Pclass_Interaction"] = df["Age"] * df["Pclass"]

print(df[
    ["FamilySize", "IsAlone", "FarePerPerson",
     "AgeGroup", "Age_Pclass_Interaction"]
].head())
```

## Question 15: Create Customer-Level Aggregation Features.

### Answer

Aggregation features summarize multiple records into meaningful group-level information.

The Titanic dataset does not contain customer transaction data, so `Ticket` can be used as a group identifier.

Useful ticket-level aggregation features include:

- Number of passengers sharing the ticket.
- Total fare for the ticket.
- Average fare for the ticket.

These features provide information about group size and fare characteristics.

```python
import pandas as pd

df = pd.read_csv("Titanic-Dataset.csv")

df["Ticket_Passenger_Count"] = df.groupby("Ticket")["PassengerId"].transform("count")
df["Ticket_Total_Fare"] = df.groupby("Ticket")["Fare"].transform("sum")
df["Ticket_Average_Fare"] = df.groupby("Ticket")["Fare"].transform("mean")

print(df[
    ["Ticket", "Ticket_Passenger_Count",
     "Ticket_Total_Fare", "Ticket_Average_Fare"]
].head(10))
```

## Question 16: Extract Features from a Date Column.

### Answer

Date columns can contain useful information such as:

- Year
- Month
- Day
- Day of Week
- Quarter
- Weekend Indicator

The Titanic dataset does not contain an actual transaction or event date.

Therefore, a synthetic date is used only to demonstrate the date feature extraction technique. It does not represent the historical Titanic event date.

In a real project, these features should be extracted from an actual date column.

```python
import pandas as pd

df = pd.read_csv("Titanic-Dataset.csv")

df["Event_Date"] = pd.date_range(
    start="1912-04-01",
    periods=len(df),
    freq="D"
)

df["Event_Year"] = df["Event_Date"].dt.year
df["Event_Month"] = df["Event_Date"].dt.month
df["Event_Day"] = df["Event_Date"].dt.day
df["Day_Of_Week"] = df["Event_Date"].dt.dayofweek
df["Weekend_Indicator"] = (df["Day_Of_Week"] >= 5).astype(int)

print(df[
    ["Event_Date", "Event_Year", "Event_Month",
     "Event_Day", "Day_Of_Week", "Weekend_Indicator"]
].head())
```

## Question 17: Create Interaction Features.

### Answer

Interaction features represent relationships between existing features.

For the Titanic dataset, useful examples are:

- `Age_Pclass_Interaction = Age × Pclass`
- `Family_Fare_Interaction = FamilySize × Fare`
- `Sibling_Parent_Difference = SibSp − Parch`

These features provide additional information about the relationship between passenger characteristics.

They should be created based on meaningful relationships rather than creating every possible combination.

```python
import pandas as pd

df = pd.read_csv("Titanic-Dataset.csv")

df["Age"] = df["Age"].fillna(df["Age"].median())
df["FamilySize"] = df["SibSp"] + df["Parch"] + 1

df["Age_Pclass_Interaction"] = df["Age"] * df["Pclass"]
df["Family_Fare_Interaction"] = df["FamilySize"] * df["Fare"]
df["Sibling_Parent_Difference"] = df["SibSp"] - df["Parch"]

print(df[
    ["Age_Pclass_Interaction",
     "Family_Fare_Interaction",
     "Sibling_Parent_Difference"]
].head())
```

## Question 18: Perform Feature Selection using Multiple Techniques.

### Answer

Feature Selection identifies useful features and removes irrelevant or redundant features.

Common techniques include:

### Variance Threshold
Removes features with very low variance.

### Correlation Analysis
Identifies relationships between numerical features and possible redundancy.

### Mutual Information
Measures the dependency between features and the target.

### RFE
Recursive Feature Elimination repeatedly removes less useful features using a Machine Learning model.

### Tree-Based Feature Importance
Uses models such as Random Forest to rank features according to their predictive usefulness.

Feature selection should be performed using training data in a real ML workflow so that the test data does not influence the selection process.

```python
import pandas as pd
from sklearn.feature_selection import VarianceThreshold, mutual_info_classif

df = pd.read_csv("Titanic-Dataset.csv")

features = ["Pclass", "Age", "SibSp", "Parch", "Fare"]

X = df[features].copy()
X[["Age", "Fare"]] = X[["Age", "Fare"]].fillna(
    X[["Age", "Fare"]].median()
)

y = df["Survived"]

# Variance
variance = pd.DataFrame({
    "Feature": X.columns,
    "Variance": X.var()
})

# Mutual Information
mi = mutual_info_classif(X, y, random_state=42)

mutual_information = pd.DataFrame({
    "Feature": X.columns,
    "Mutual_Information": mi
}).sort_values("Mutual_Information", ascending=False)

print("Variance:")
print(variance)

print("\nMutual Information:")
print(mutual_information)
```
