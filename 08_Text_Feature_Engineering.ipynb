```python
# Notebook 8 – Text Feature Engineering

This notebook covers how raw text can be converted into numeric features that a Machine Learning model can actually use.
```

```python
import pandas as pd
df = pd.read_csv('data.csv', encoding='latin1')
df.head()
```

## 1. Text Length

**Definition:** The total number of characters in a piece of text, including spaces.

**Example:** "Hello World" has a length of 11.

**Why it is used?** Longer descriptions or reviews can carry more information, so text length is a simple signal a model can learn from.

```python
df['Description'] = df['Description'].fillna('').astype(str)
df['Text_Length'] = df['Description'].apply(len)
df[['Description', 'Text_Length']].head()
```

**Code Explanation:** The Description column is converted to string type to avoid errors, then len is applied to each row to count all characters.

## 2. Character Count

**Definition:** Similar to text length, but often used to specifically mean the count of all characters excluding or including spaces depending on the use case.

**Example:** "Hi there" has 8 characters with spaces, 7 without.

**Why it is used?** Helps compare how detailed or short different text entries are.

```python
df['Char_Count'] = df['Description'].apply(lambda x: len(x.replace(' ', '')))
df[['Description', 'Char_Count']].head()
```

**Code Explanation:** Spaces are removed from each description, then the remaining characters are counted using len.

## 3. Word Count

**Definition:** The number of words present in a piece of text.

**Example:** "Red Woolly Hottie White Heart" has 5 words.

**Why it is used?** Word count can hint at how descriptive or detailed a text is, which can matter for tasks like spam detection or content quality.

```python
df['Word_Count'] = df['Description'].apply(lambda x: len(x.split()))
df[['Description', 'Word_Count']].head()
```

**Code Explanation:** split breaks the text into a list of words using spaces, and len counts how many words are in that list.

## 4. Sentence Count

**Definition:** The number of sentences in a piece of text, usually detected using punctuation like periods, question marks, or exclamation marks.

**Example:** "This is great. I love it!" has 2 sentences.

**Why it is used?** Helps understand how structured or long-form a piece of text is, useful for reviews, comments, or articles.

```python
df['Sentence_Count'] = df['Description'].apply(lambda x: len([s for s in x.split('.') if s.strip() != '']))
df[['Description', 'Sentence_Count']].head()
```

**Code Explanation:** The text is split wherever a period appears, empty pieces are ignored, and the remaining pieces are counted as sentences.

## 5. Number of Digits

**Definition:** The count of numeric characters (0 to 9) present in the text.

**Example:** "Box 22623" has 5 digits.

**Why it is used?** Useful for detecting product codes, prices, or identifiers mixed into text fields.

```python
df['Digit_Count'] = df['Description'].apply(lambda x: sum(c.isdigit() for c in x))
df[['Description', 'Digit_Count']].head()
```

**Code Explanation:** Each character in the text is checked with isdigit, and the total count of digit characters is summed up.

## 6. Number of Special Characters

**Definition:** The count of characters that are not letters, digits, or spaces, like punctuation and symbols.

**Example:** "Poppy's Playhouse!" has 2 special characters, the apostrophe and the exclamation mark.

**Why it is used?** Helps flag unusual formatting, spam-like text, or noisy data entries.

```python
df['Special_Char_Count'] = df['Description'].apply(lambda x: sum(not c.isalnum() and not c.isspace() for c in x))
df[['Description', 'Special_Char_Count']].head()
```

**Code Explanation:** Each character is checked to see if it is not alphanumeric and not a space, and those characters are counted as special characters.

## 7. Uppercase Count

**Definition:** The number of uppercase letters present in the text.

**Example:** "WHITE HANGING HEART" has all uppercase letters.

**Why it is used?** High uppercase usage can indicate emphasis, shouting in text, or all-caps product names.

```python
df['Uppercase_Count'] = df['Description'].apply(lambda x: sum(c.isupper() for c in x))
df[['Description', 'Uppercase_Count']].head()
```

**Code Explanation:** Each character is checked with isupper, and the total number of uppercase letters is counted.

## 8. Lowercase Count

**Definition:** The number of lowercase letters present in the text.

**Example:** "Poppy's Playhouse Bedroom" has several lowercase letters.

**Why it is used?** Balances with uppercase count to understand the overall casing style of the text.

```python
df['Lowercase_Count'] = df['Description'].apply(lambda x: sum(c.islower() for c in x))
df[['Description', 'Lowercase_Count']].head()
```

**Code Explanation:** Each character is checked with islower, and the total number of lowercase letters is counted.

## 9. Keyword Count

**Definition:** The number of times specific target words appear in the text.

**Example:** Counting how many times the word "box" appears across product descriptions.

**Why it is used?** Useful when certain words are strong indicators of a category, like detecting gift items or fragile items.

```python
keyword = 'box'
df['Keyword_Count'] = df['Description'].apply(lambda x: x.lower().count(keyword))
df[['Description', 'Keyword_Count']].head()
```

**Code Explanation:** The text is converted to lowercase so the search is not case sensitive, then count checks how many times the keyword appears.

## 10. Bag of Words

**Definition:** A method that converts text into a matrix of word counts, where each column represents a unique word and each row represents a document.

**Example:** "red box" and "blue box" become two rows sharing the word "box" as a common column.

**Why it is used?** Turns raw text into numeric features that models can process, based purely on how often each word appears.

```python
from sklearn.feature_extraction.text import CountVectorizer
vectorizer = CountVectorizer(max_features=20)
bow_matrix = vectorizer.fit_transform(df['Description'])
bow_df = pd.DataFrame(bow_matrix.toarray(), columns=vectorizer.get_feature_names_out())
bow_df.head()
```

**Code Explanation:** CountVectorizer builds a vocabulary of the top 20 words, then converts every description into a row of word counts based on that vocabulary.

## 11. TF-IDF

**Definition:** Term Frequency-Inverse Document Frequency, a score that shows how important a word is to a document compared to the whole dataset.

**Example:** Common words like "the" get a low score, while rare but meaningful words get a high score.

**Why it is used?** Reduces the weight of overly common words and highlights words that actually carry meaning.

```python
from sklearn.feature_extraction.text import TfidfVectorizer
tfidf_vectorizer = TfidfVectorizer(max_features=20)
tfidf_matrix = tfidf_vectorizer.fit_transform(df['Description'])
tfidf_df = pd.DataFrame(tfidf_matrix.toarray(), columns=tfidf_vectorizer.get_feature_names_out())
tfidf_df.head()
```

**Code Explanation:** TfidfVectorizer builds the same kind of word matrix as CountVectorizer, but weighs each word by how unique and meaningful it is across all descriptions.

## 12. N-Grams

**Definition:** Groups of N consecutive words treated as a single unit, instead of looking at words one at a time.

**Example:** For "red heart box", the bigrams (2-grams) are "red heart" and "heart box".

**Why it is used?** Captures word order and short phrases, which single word counts miss completely.

```python
ngram_vectorizer = CountVectorizer(ngram_range=(2, 2), max_features=20)
ngram_matrix = ngram_vectorizer.fit_transform(df['Description'])
ngram_df = pd.DataFrame(ngram_matrix.toarray(), columns=ngram_vectorizer.get_feature_names_out())
ngram_df.head()
```

**Code Explanation:** ngram_range is set to (2, 2) so the vectorizer captures pairs of consecutive words instead of single words, then builds a count matrix from those word pairs.

## How Raw Text Becomes Machine Learning Features

Raw text on its own means nothing to a Machine Learning model, since models only understand numbers. The features built above turn text into numbers in two main ways.

**Structural features** like text length, word count, digit count, and uppercase count describe the shape and style of the text without caring about its actual meaning.

**Content-based features** like Bag of Words, TF-IDF, and N-Grams describe what the text actually says, by turning words and phrases into numeric columns a model can learn patterns from.

Together, these features let a model use text data the same way it would use any other numeric column.
